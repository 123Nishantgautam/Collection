```python

# E-Commerce Customer Segmentation using RFM
# Algorithms: K-Means, DBSCAN, PCA and t-SNE

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics import silhouette_score


# STEP 1: SETTINGS

RAW_FILE = "Copy of Online Retail110.csv"
INPUT_FILE = "rfm_customers.csv"
OUTPUT_FILE = "customer_segments.csv"
PLOT_DIR = "plots"

RANDOM_STATE = 42

os.makedirs(PLOT_DIR, exist_ok=True)


# STEP 2: LOAD ORIGINAL DATASET

df_raw = pd.read_csv(RAW_FILE, encoding="latin1")
df_raw.columns = df_raw.columns.str.strip()

print("Original dataset loaded")
print(df_raw.head())
print("Dataset shape:", df_raw.shape)

required_columns = [
    "InvoiceNo", "InvoiceDate", "Quantity",
    "UnitPrice", "CustomerID"
]

missing = [
    col for col in required_columns
    if col not in df_raw.columns
]

if missing:
    raise ValueError(f"Missing columns: {missing}")


# STEP 3: CLEAN DATA

# Remove records without CustomerID
df_raw = df_raw.dropna(subset=["CustomerID"]).copy()

# Remove cancelled invoices
df_raw = df_raw[
    ~df_raw["InvoiceNo"].astype(str).str.startswith("C")
].copy()

# Convert columns into correct data types
df_raw["Quantity"] = pd.to_numeric(
    df_raw["Quantity"], errors="coerce"
)

df_raw["UnitPrice"] = pd.to_numeric(
    df_raw["UnitPrice"], errors="coerce"
)

df_raw["InvoiceDate"] = pd.to_datetime(
    df_raw["InvoiceDate"], errors="coerce"
)

# Remove invalid values
df_raw = df_raw.dropna(
    subset=["InvoiceDate", "Quantity", "UnitPrice"]
)

df_raw = df_raw[
    (df_raw["Quantity"] > 0) &
    (df_raw["UnitPrice"] > 0)
].copy()

if df_raw.empty:
    raise ValueError("No valid transaction data found.")


# STEP 4: CREATE RFM DATASET

# Calculate total amount for each transaction
df_raw["TotalAmount"] = (
    df_raw["Quantity"] * df_raw["UnitPrice"]
)

# Reference date: one day after the latest transaction
reference_date = (
    df_raw["InvoiceDate"].max() + pd.Timedelta(days=1)
)

# Calculate Recency, Frequency and Monetary
df = df_raw.groupby("CustomerID").agg(
    Recency=(
        "InvoiceDate",
        lambda dates: (reference_date - dates.max()).days
    ),
    Frequency=("InvoiceNo", "nunique"),
    Monetary=("TotalAmount", "sum")
).reset_index()

df["CustomerID"] = df["CustomerID"].astype(str)

# Save RFM dataset
df.to_csv(INPUT_FILE, index=False)

print("\nRFM dataset created:", INPUT_FILE)
print(df.head())
print("Total customers:", len(df))


# STEP 5: VALIDATE RFM DATA

features = ["Recency", "Frequency", "Monetary"]

for col in features:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=features)
df = df.drop_duplicates(subset=["CustomerID"])

df = df[
    (df["Recency"] >= 0) &
    (df["Frequency"] >= 0) &
    (df["Monetary"] >= 0)
].copy()

if len(df) < 4:
    raise ValueError("At least 4 customers are required.")

if df[features].nunique().min() < 2:
    raise ValueError(
        "Each RFM feature must have at least two distinct values."
    )

print("\nClean RFM data:")
print(df.head())
print("Shape:", df.shape)


# STEP 6: FEATURE SCALING

X = df[features].copy()

# Reduce skew in the RFM values
X_log = np.log1p(X)

# Scale all features to comparable ranges
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_log)


# STEP 7: K-MEANS CLUSTERING

kmeans = KMeans(
    n_clusters=4,
    random_state=RANDOM_STATE,
    n_init=10
)

df["KMeans_Cluster"] = kmeans.fit_predict(X_scaled)

print("\nK-Means cluster counts:")
print(df["KMeans_Cluster"].value_counts().sort_index())


# STEP 8: DBSCAN CLUSTERING

dbscan = DBSCAN(
    eps=0.8,
    min_samples=10
)

df["DBSCAN_Cluster"] = dbscan.fit_predict(X_scaled)

print("\nDBSCAN cluster counts:")
print(df["DBSCAN_Cluster"].value_counts().sort_index())


# STEP 9: PCA DIMENSION REDUCTION

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

df["PCA1"] = X_pca[:, 0]
df["PCA2"] = X_pca[:, 1]

print("\nPCA explained variance:")
print(pca.explained_variance_ratio_)

print(
    "Total variance explained:",
    round(pca.explained_variance_ratio_.sum() * 100, 2),
    "%"
)


# STEP 10: t-SNE VISUALIZATION

perplexity = min(30, max(2, (len(df) - 1) // 3))

tsne = TSNE(
    n_components=2,
    perplexity=perplexity,
    random_state=RANDOM_STATE,
    init="pca",
    learning_rate="auto"
)

X_tsne = tsne.fit_transform(X_scaled)

df["TSNE1"] = X_tsne[:, 0]
df["TSNE2"] = X_tsne[:, 1]


# STEP 11: K-MEANS PCA GRAPH

plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    df["PCA1"],
    df["PCA2"],
    c=df["KMeans_Cluster"],
    cmap="tab10",
    s=35,
    alpha=0.8
)

plt.title("K-Means Customer Segments (PCA)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.colorbar(scatter, label="K-Means Cluster")
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "pca_kmeans.png"))
plt.show()
plt.close()


# STEP 12: DBSCAN PCA GRAPH

plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    df["PCA1"],
    df["PCA2"],
    c=df["DBSCAN_Cluster"],
    cmap="tab10",
    s=35,
    alpha=0.8
)

plt.title("DBSCAN Customer Segments (PCA)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.colorbar(scatter, label="DBSCAN Cluster (-1 = Noise)")
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "pca_dbscan.png"))
plt.show()
plt.close()


# STEP 13: K-MEANS t-SNE GRAPH

plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    df["TSNE1"],
    df["TSNE2"],
    c=df["KMeans_Cluster"],
    cmap="tab10",
    s=35,
    alpha=0.8
)

plt.title("K-Means Customer Segments (t-SNE)")
plt.xlabel("t-SNE Dimension 1")
plt.ylabel("t-SNE Dimension 2")
plt.colorbar(scatter, label="K-Means Cluster")
plt.tight_layout()
plt.savefig(os.path.join(PLOT_DIR, "tsne_kmeans.png"))
plt.show()
plt.close()


# STEP 14: CUSTOMER SEGMENT SUMMARY

summary = df.groupby("KMeans_Cluster").agg(
    Customers=("CustomerID", "count"),
    Avg_Recency=("Recency", "mean"),
    Avg_Frequency=("Frequency", "mean"),
    Avg_Monetary=("Monetary", "mean"),
    Median_Recency=("Recency", "median"),
    Median_Frequency=("Frequency", "median"),
    Median_Monetary=("Monetary", "median")
).round(2)

print("\nK-Means RFM summary:")
print(summary)


# STEP 15: MARKETING PERSONAS

overall_recency = df["Recency"].median()
overall_frequency = df["Frequency"].median()
overall_monetary = df["Monetary"].median()


def assign_persona(row):

    recent = row["Avg_Recency"] <= overall_recency
    frequent = row["Avg_Frequency"] >= overall_frequency
    high_spend = row["Avg_Monetary"] >= overall_monetary

    if recent and frequent and high_spend:
        return (
            "VIP / Loyal Customers",
            "Reward loyalty and offer premium products."
        )

    if recent and frequent:
        return (
            "Regular Active Customers",
            "Offer loyalty points and product bundles."
        )

    if recent and not frequent:
        return (
            "New / Promising Customers",
            "Encourage a second purchase with an offer."
        )

    if not recent and frequent:
        return (
            "At-Risk Regular Customers",
            "Send personalized win-back offers."
        )

    if high_spend:
        return (
            "High-Spending Customers",
            "Offer premium recommendations and VIP benefits."
        )

    return (
        "Low-Engagement Customers",
        "Try reactivation campaigns and relevant discounts."
    )


persona_rows = []

for cluster_id, row in summary.iterrows():

    persona, strategy = assign_persona(row)

    persona_rows.append({
        "KMeans_Cluster": cluster_id,
        "Persona": persona,
        "Marketing_Strategy": strategy
    })

personas = pd.DataFrame(persona_rows)

print("\nMarketing personas:")
print(personas.to_string(index=False))

personas.to_csv("customer_personas.csv", index=False)


# STEP 16: SILHOUETTE SCORES

kmeans_labels = df["KMeans_Cluster"].to_numpy()

if 1 < len(np.unique(kmeans_labels)) < len(kmeans_labels):
    score = silhouette_score(X_scaled, kmeans_labels)
    print("\nK-Means Silhouette Score:", round(score, 3))
else:
    print("\nK-Means silhouette score skipped.")

# Ignore DBSCAN noise points (-1)
db_labels = df["DBSCAN_Cluster"].to_numpy()
non_noise = db_labels != -1
valid_labels = db_labels[non_noise]

if (
    len(valid_labels) > 2 and
    1 < len(np.unique(valid_labels)) < len(valid_labels)
):
    score = silhouette_score(
        X_scaled[non_noise],
        valid_labels
    )
    print("DBSCAN Silhouette Score:", round(score, 3))
else:
    print("DBSCAN silhouette score skipped.")


# STEP 17: SAVE FINAL RESULTS

df.to_csv(OUTPUT_FILE, index=False)

print("\nRFM dataset saved as:", INPUT_FILE)
print("Final segments saved as:", OUTPUT_FILE)
print("Marketing personas saved as: customer_personas.csv")
print("Graphs saved in:", PLOT_DIR)
print("\nCustomer segmentation completed successfully!")
```
